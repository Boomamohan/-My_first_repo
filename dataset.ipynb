```python
days = ["Mon","Tue","Wed","Thur","Fri","Sat","Sun"]
sales = [20,92,320,186,75,89,65]

plt.plot(days,sales,marker = "o",color = "black",linestyle = "dashed",markersize=10,markerfacecolor="green",markeredgecolor="yellow")
plt.xlabel("Weekdays")
plt.ylabel("Sales")
plt.title("Plot for Days vs Sales")
for i in range(len(days)):
  plt.text(days[i],sales[i]+20,(sales[i]),ha ='center',va = 'bottom')

plt.ylim(0,400)
plt.gca().set_facecolor('lightblue')
plt.grid()
plt.show()
```

```python
days = ["Mon","Tue","Wed","Thur","Fri","Sat","Sun"]
sales_Mango = [50,20,120,140,75,150,300]
sales_Apple = [20,72,220,86,175,49,55]
sales_Grapes = [120,220,80,68,270,99,150]

plt.plot(days,sales_Mango,marker = "o",color = "black",linestyle = "dashed",markersize=10,markerfacecolor="green",markeredgecolor="yellow")
plt.plot(days,sales_Apple,marker = "o",color = "black",linestyle = "solid",markersize=10,markerfacecolor="red",markeredgecolor="yellow")
plt.plot(days,sales_Grapes,marker = "o",color = "black",linestyle = "dotted",markersize=10,markerfacecolor="blue",markeredgecolor="yellow")
plt.xlabel("Weekdays")
plt.ylabel("Sales")
plt.title("Plot for Days vs Sales")
for i in range(len(days)):
  plt.text(days[i],sales_Mango[i]+20,(sales_Mango[i]),ha ='center',va = 'bottom')
  plt.text(days[i],sales_Apple[i]+20,(sales_Apple[i]),ha ='center',va = 'bottom')
  plt.text(days[i],sales_Grapes[i]+20,(sales_Grapes[i]),ha ='center',va = 'bottom')

plt.ylim(0,350)
plt.gca().set_facecolor('lightblue')
plt.grid()
plt.show()
```
